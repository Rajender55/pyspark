```python
# 2. to read .csv file which already has schema

df1 = spark.read.csv("/home/rajender/pyspark/pyspark/sample_employees.csv", header=True, inferSchema=True)
df1.show()
```

```python
# to see the schema of the columns available
df1.printSchema()
```

```python
# 3. how to filter rows based on seperate condition

df_filter = df1.filter((df1.age>30) & (df1.gender == "Male"))
df_filter.show()
```

```python
# 4. what is the average salary of the each department and sort them in desc order of avg salary
df_avg_salary = df1.groupBy("department").agg({"salary": "avg"})
df_avg_salary.show()
```

```python
# sort the avg_salary in desc order
df_avg_salary.sort(desc("avg(salary)")).show()
```

```python
# 5. count the nulls in a dataframe and handle them accordingly
df_null = df1.select([sum(when(col(c).isNull(), 1).otherwise(0)).alias(c) for c in df1.columns])
df_null.show()
```

```python
df1.count()
```

```python
#filling null values
# for the age column, we can fill the nulls with average age of the employees
avg_age = df1.select(avg("age")).first()[0]
df1=df1.na.fill({"age": avg_age})

# and for the country column, we can drop the rows with empty values
df1= df1.na.drop()
df1.count()
```

```python
#6. How to use window functions in pyspark
from pyspark.sql.window import Window
window_spec = Window.partitionBy("department").orderBy(df1.salary.desc())
df1=df1.withColumn("rank", rank().over(window_spec))
df1.select("id", "first_name", "department", "salary", "rank").show()
```

```python
#8. How to write DataFrame to Parquet with partitioning?

df1.write.mode("overwrite").partitionBy("department").parquet("parquet_files")
```

```python
#9. How to register a DataFrame as a temporary SQL table and query it?
df1.createOrReplaceTempView("datatable")
spark.sql("select * from datatable where age>30").show()
```

```python
#10 How to apply a UDF (User Defined Function)?
from pyspark.sql.types import *

def name_in_uppercase(name):
    return name.upper()

udf_name = udf(name_in_uppercase, StringType())
df1 = df1.withColumn("FIRST_NAME", udf_name(df1.first_name))
df1.show(5)
```

```python
df1.count()
```

```python
#11. top 2 persons in each department with greatest salary
window1=Window.partitionBy("department").orderBy(df1.salary.desc())
df_top3 = df1.withColumn("rnk", row_number().over(window1)).filter("rnk<=3")
df_top3.show()
```
