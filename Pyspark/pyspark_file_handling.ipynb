```python
sc = spark.sparkContext
```

```python
state_names = {"CA":"California", "TX":"Texas", "WA":"Washington"}
broardcast_variable = sc.broadcast(state_names)
rdd = sc.parallelize([("Alice", "CA"), ("Bob", "TX"), ("Eve", "WA")])

rdd.collect()
```

```python
#add state name to the data
def add_state_name(record):
    name, state_name = record 
    state_full_name = broardcast_variable.value.get(state_name, "Unknown")
    return (name, state_name, state_full_name)

rdd.map(add_state_name).collect() 
```
